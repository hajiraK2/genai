```python
!pip install -q gensim transformers torch

import gensim.downloader as api
from transformers import pipeline

word_vectors = api.load("glove-wiki-gigaword-100")

generator = pipeline(
    "text-generation",
    model="gpt2"
)

prompt = "Explain machine learning in healthcare"

def enrich_prompt(prompt):

    enriched_words = []

    for word in prompt.split():

        enriched_words.append(word)

        if word in word_vectors:

            similar_words = word_vectors.most_similar(
                word,
                topn=2
            )

            enriched_words.append(similar_words[0][0])
            enriched_words.append(similar_words[1][0])

    return " ".join(enriched_words)

enriched_prompt = enrich_prompt(prompt)

original_output = generator(
    prompt,
    max_length=100
)[0]["generated_text"]

enriched_output = generator(
    enriched_prompt,
    max_length=100
)[0]["generated_text"]

print("Original Prompt:")
print(prompt)

print("\nEnriched Prompt:")
print(enriched_prompt)

print("\nOutput (Original):")
print(original_output)

print("\nOutput (Enriched):")
print(enriched_output)

print("\nComparison:")
print("Original length:", len(original_output.split()))
print("Enriched length:", len(enriched_output.split()))

print("Insight: Enriched prompt may provide more context but can also introduce noise.")
```
