```python
!pip install -q gensim

import random
import gensim.downloader as api

model = api.load("word2vec-google-news-300")

def get_similar_words(seed_word):

    try:
        similar_words = model.most_similar(
            seed_word,
            topn=5
        )

        return [word for word, score in similar_words]

    except KeyError:

        return [seed_word]

def generate_paragraph(seed_word):

    words = get_similar_words(seed_word)

    random.shuffle(words)

    paragraph = (
        f"Once upon a time, there was a {seed_word} "
        f"who dreamed of adventure. "
    )

    for word in words:

        paragraph += (
            f"One day, it stumbled upon a {word}, "
            f"leading it into an unexpected journey. "
        )

    paragraph += (
        "The experience changed its life forever."
    )

    return paragraph

seed_word = input("Enter a seed word: ")

print("\nGenerated Paragraph:\n")

print(generate_paragraph(seed_word))
```
